```python
def tokenize_and_align_labels(examples):
    tokenized = tokenizer(
        examples["tokens"],
        truncation=True,
        is_split_into_words=True
    )

    all_labels = []

    for i, labels in enumerate(examples["ner_tags"]):
        word_ids = tokenized.word_ids(batch_index=i)
        previous_word_id = None
        label_ids = []

        for word_id in word_ids:
            if word_id is None:
                label_ids.append(-100)
            elif word_id != previous_word_id:
                label_ids.append(labels[word_id])
            else:
                label_ids.append(-100)

            previous_word_id = word_id

        all_labels.append(label_ids)

    tokenized["labels"] = all_labels
    return tokenized


tokenized_dataset = dataset.map(
    tokenize_and_align_labels,
    batched=True
)

print(tokenized_dataset)
```

```python
def convert_to_bio(resume):
    text = resume["content"]
    annotations = []

    label_map = {
        "Name": "NAME",
        "Email Address": "EMAIL",
        "Location": "LOCATION",
        "Companies worked at": "COMPANY",
        "Designation": "DESIGNATION",
        "College Name": "COLLEGE",
        "Degree": "DEGREE",
        "Graduation Year": "GRADUATION_YEAR",
        "Skills": "SKILL",
        "Years of Experience": "EXPERIENCE"
    }

    for ann in resume["annotation"]:
        if not ann.get("label"):
            continue

        original_label = ann["label"][0]

        if original_label not in label_map:
            continue

        label = label_map[original_label]

        for point in ann.get("points", []):
            start = point["start"]
            end = point["end"]

            annotations.append((start, end, label))

    annotations.sort(key=lambda x: x[0])

    words = []
    current_word = ""
    word_start = None

    for i, char in enumerate(text):
        if char.isspace():
            if current_word:
                words.append((current_word, word_start, i))
                current_word = ""
                word_start = None
        else:
            if current_word == "":
                word_start = i
            current_word += char

    if current_word:
        words.append((current_word, word_start, len(text)))

    bio_labels = []

    for word, start, end in words:
        label = "O"

        for ent_start, ent_end, ent_label in annotations:
            if start >= ent_start and end <= ent_end:
                if start == ent_start:
                    label = "B-" + ent_label
                else:
                    label = "I-" + ent_label
                break

        bio_labels.append(label)

    return {
        "tokens": [w[0] for w in words],
        "ner_tags": [label2id[x] for x in bio_labels]
    }


bio_data = [convert_to_bio(resume) for resume in data]

print("Converted resumes:", len(bio_data))
print("\nFirst example:")
print(bio_data[0]["tokens"][:30])
print(bio_data[0]["ner_tags"][:30])
```

```python
# ============================================================
# FINAL CLEAN NER DATA PREPARATION
# ============================================================

import json
import re
from collections import Counter
from sklearn.model_selection import train_test_split
from datasets import Dataset, DatasetDict

# ------------------------------------------------------------
# 1. LABELS
# ------------------------------------------------------------

ENTITY_TYPES = [
    "NAME",
    "LOCATION",
    "COMPANY",
    "DESIGNATION",
    "COLLEGE",
    "DEGREE",
    "GRADUATION_YEAR",
    "SKILL",
    "EXPERIENCE"
]

NER_LABELS = ["O"]

for entity in ENTITY_TYPES:
    NER_LABELS.append("B-" + entity)
    NER_LABELS.append("I-" + entity)

label2id = {label: i for i, label in enumerate(NER_LABELS)}
id2label = {i: label for i, label in enumerate(NER_LABELS)}

print("BIO labels:", len(NER_LABELS))
print(NER_LABELS)


# ------------------------------------------------------------
# 2. LOAD REAL DATATURKS DATA
# ------------------------------------------------------------

data = []

with open(
    "Entity Recognition in Resumes.json",
    "r",
    encoding="utf-8"
) as f:

    for line in f:
        line = line.strip()

        if line:
            data.append(json.loads(line))

print("Resumes loaded:", len(data))


# ------------------------------------------------------------
# 3. DATASET LABEL MAPPING
# ------------------------------------------------------------

source_to_target = {
    "Name": "NAME",
    "Location": "LOCATION",
    "Companies worked at": "COMPANY",
    "Designation": "DESIGNATION",
    "College Name": "COLLEGE",
    "Degree": "DEGREE",
    "Graduation Year": "GRADUATION_YEAR",
    "Skills": "SKILL",
    "Years of Experience": "EXPERIENCE"
}


# ------------------------------------------------------------
# 4. CLEAN CHARACTER SPANS
# ------------------------------------------------------------

def get_clean_annotations(resume):

    text = resume["content"]
    annotations = []

    for ann in resume.get("annotation", []):

        labels = ann.get("label", [])

        if not labels:
            continue

        source_label = labels[0]

        if source_label not in source_to_target:
            continue

        target_label = source_to_target[source_label]

        for point in ann.get("points", []):

            start = point.get("start")
            end = point.get("end")

            if start is None or end is None:
                continue

            # Safety
            start = max(0, int(start))
            end = min(len(text), int(end))

            if start >= end:
                continue

            entity_text = text[start:end]

            # Ignore empty/whitespace annotations
            if not entity_text.strip():
                continue

            annotations.append(
                (start, end, target_label)
            )

    # Remove exact duplicates
    annotations = list(set(annotations))

    # Sort by start position and then longest span first
    annotations.sort(
        key=lambda x: (x[0], -(x[1] - x[0]))
    )

    # --------------------------------------------------------
    # Remove conflicting overlapping annotations.
    # Keep the longest annotation when spans overlap.
    # --------------------------------------------------------

    cleaned = []

    for candidate in annotations:

        start, end, label = candidate

        overlaps = False

        for existing_start, existing_end, existing_label in cleaned:

            if (
                start < existing_end
                and end > existing_start
            ):
                overlaps = True
                break

        if not overlaps:
            cleaned.append(candidate)

    cleaned.sort(key=lambda x: x[0])

    return cleaned


# ------------------------------------------------------------
# 5. CHARACTER SPANS → WORD-LEVEL BIO
# ------------------------------------------------------------

def resume_to_bio(resume):

    text = resume["content"]

    annotations = get_clean_annotations(resume)

    # --------------------------------------------------------
    # Proper whitespace word spans
    # --------------------------------------------------------

    words = []

    for match in re.finditer(r"\S+", text):

        words.append(
            (
                match.group(),
                match.start(),
                match.end()
            )
        )

    tokens = []
    tags = []

    for word, word_start, word_end in words:

        selected_label = None
        is_beginning = False

        # Find entity containing this word
        for ent_start, ent_end, ent_label in annotations:

            # Word overlaps entity
            if (
                word_start < ent_end
                and word_end > ent_start
            ):

                selected_label = ent_label

                # Word starts at entity beginning
                if word_start <= ent_start:
                    is_beginning = True

                break

        tokens.append(word)

        if selected_label is None:

            tags.append("O")

        else:

            if is_beginning:
                tags.append("B-" + selected_label)
            else:
                tags.append("I-" + selected_label)

    return {
        "tokens": tokens,
        "ner_tags": [
            label2id[tag]
            for tag in tags
        ]
    }


# ------------------------------------------------------------
# 6. CONVERT ALL RESUMES
# ------------------------------------------------------------

bio_data = []

for resume in data:

    converted = resume_to_bio(resume)

    if len(converted["tokens"]) > 0:
        bio_data.append(converted)

print("BIO resumes:", len(bio_data))


# ------------------------------------------------------------
# 7. CHECK LABEL DISTRIBUTION
# ------------------------------------------------------------

counts = Counter()

for example in bio_data:

    for tag_id in example["ner_tags"]:

        counts[id2label[tag_id]] += 1

print("\nLABEL DISTRIBUTION")
print("=" * 40)

for label, count in counts.most_common():

    print(f"{label:25s} {count}")


# ------------------------------------------------------------
# 8. SPLIT
# ------------------------------------------------------------

train_data, temp_data = train_test_split(
    bio_data,
    test_size=0.20,
    random_state=42
)

val_data, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    random_state=42
)

print("\nDATA SPLIT")
print("=" * 40)
print("Training   :", len(train_data))
print("Validation :", len(val_data))
print("Test       :", len(test_data))


# ------------------------------------------------------------
# 9. HUGGING FACE DATASETS
# ------------------------------------------------------------

ner_dataset = DatasetDict({

    "train": Dataset.from_list(train_data),

    "validation": Dataset.from_list(val_data),

    "test": Dataset.from_list(test_data)
})


# ------------------------------------------------------------
# 10. SANITY CHECK
# ------------------------------------------------------------

print("\nSANITY CHECK")
print("=" * 40)

example = train_data[0]

for token, tag_id in zip(
    example["tokens"],
    example["ner_tags"]
):

    tag = id2label[tag_id]

    if tag != "O":

        print(
            f"{token:30s} -> {tag}"
        )

print("\nFINAL DATA PREPARATION COMPLETE")
```

```python
import numpy as np
from collections import defaultdict

# Get predictions on unseen test set
pred_output = trainer.predict(tokenized_ner_dataset["test"])

predictions = np.argmax(pred_output.predictions, axis=2)
true_labels = pred_output.label_ids

# Collect entity-level token statistics
stats = defaultdict(lambda: {"tp": 0, "fp": 0, "fn": 0})

for pred_seq, true_seq in zip(predictions, true_labels):

    for pred_id, true_id in zip(pred_seq, true_seq):

        # Ignore special/padding tokens
        if true_id == -100:
            continue

        pred_label = id2label[pred_id]
        true_label = id2label[true_id]

        # Ignore O for per-entity statistics
        if true_label != "O":

            if pred_label == true_label:
                stats[true_label]["tp"] += 1
            else:
                stats[true_label]["fn"] += 1
                if pred_label != "O":
                    stats[pred_label]["fp"] += 1

        elif pred_label != "O":
            stats[pred_label]["fp"] += 1


print("==============================================")
print("NER TEST SET EVALUATION")
print("==============================================")

total_tp = total_fp = total_fn = 0

for label in NER_LABELS:

    if label == "O":
        continue

    tp = stats[label]["tp"]
    fp = stats[label]["fp"]
    fn = stats[label]["fn"]

    precision = tp / (tp + fp) if (tp + fp) else 0
    recall = tp / (tp + fn) if (tp + fn) else 0
    f1 = (
        2 * precision * recall / (precision + recall)
        if (precision + recall)
        else 0
    )

    total_tp += tp
    total_fp += fp
    total_fn += fn

    print(
        f"{label:20} "
        f"Precision: {precision:.3f}  "
        f"Recall: {recall:.3f}  "
        f"F1: {f1:.3f}  "
        f"(TP={tp}, FP={fp}, FN={fn})"
    )

# Overall micro scores
precision = total_tp / (total_tp + total_fp) if (total_tp + total_fp) else 0
recall = total_tp / (total_tp + total_fn) if (total_tp + total_fn) else 0
f1 = (
    2 * precision * recall / (precision + recall)
    if (precision + recall)
    else 0
)

print("\n==============================================")
print("OVERALL TOKEN-LEVEL RESULTS")
print("==============================================")
print(f"Precision : {precision:.3f}")
print(f"Recall    : {recall:.3f}")
print(f"F1        : {f1:.3f}")
print("==============================================")
```
